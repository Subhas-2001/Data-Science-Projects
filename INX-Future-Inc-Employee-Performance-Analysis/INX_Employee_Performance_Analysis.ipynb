## EDA Findings and Their Impact on Preprocessing & Model Building:

- **Performance Rating is imbalanced**, with Rating 3 being the dominant class.

  **Impact:** Class imbalance should be considered during model training and evaluation. Stratified splitting and suitable metrics such as F1-score, precision, recall, and confusion matrix should be used.

- **Categorical features contain multiple categories**, particularly EmpJobRole and EmpDepartment.

  **Impact:** Categorical encoding will be required before applying machine-learning algorithms. Appropriate encoding should be selected based on the  model and feature characteristics.

- **Some categorical categories have very low representation**, such as certain job roles and departments.

  **Impact:** Rare categories should be reviewed carefully during preprocessing to avoid unstable model patterns and unreliable predictions.

- **Numerical features have different scales and ranges.** For example, Age, EmpHourlyRate, and experience-related variables have substantially different ranges.

  **Impact:** Feature scaling should be considered for scale-sensitive algorithms such as Logistic Regression, KNN, and SVM, while tree-based models generally do not require scaling.

- **Several numerical variables are positively or negatively skewed**, particularly experience and promotion-related variables.

  **Impact:** Skewness should be evaluated before modeling. Transformation may be considered where appropriate, but variables should not be modified unnecessarily.

- **Some numerical variables show correlations with PerformanceRating**, especially EmpEnvironmentSatisfaction and EmpLastSalaryHikePercent.

  **Impact:** These variables should be retained and given attention during feature analysis and model interpretation. However, feature selection should not rely only on correlation.

- **Some predictor variables may be related to each other**, particularly experience-related variables such as TotalWorkExperienceInYears, ExperienceYearsAtThisCompany, and ExperienceYearsInCurrentRole.

## 15.3 Consolidate One-Hot Encoded Features:

```python
def get_original_feature(feature_name):
    
    feature_name = feature_name.replace('num__', '')
    feature_name = feature_name.replace('cat__', '')
    
    for feature in categorical_features:
        if feature_name.startswith(feature + '_'):
            return feature
    
    return feature_name


feature_importance['Original Feature'] = (
    feature_importance['Feature']
    .apply(get_original_feature)
)

business_feature_importance = (
    feature_importance
    .groupby('Original Feature', as_index=False)['Importance (%)']
    .sum()
    .sort_values(
        by='Importance (%)',
        ascending=False
    )
    .reset_index(drop=True)
)

business_feature_importance
```

# 19. Challenges Faced & Techniques Used:

**1. Imbalanced Target Variable**

**Challenge:** The PerformanceRating target variable was imbalanced, with PerformanceRating 3 accounting for the majority of observations compared with Ratings 2 and 4. This could affect the model's ability to learn the minority classes effectively.

**Technique Used:** Stratified train-test splitting was applied to maintain a similar distribution of PerformanceRating classes in both training and testing datasets. Multiple evaluation metrics were also considered instead of relying only on accuracy.

---

**2. Mixed Numerical and Categorical Features**

**Challenge:** The dataset contained both numerical and categorical variables, which cannot be directly processed in the same way by most machine-learning algorithms.

**Technique Used:** Features were segregated into numerical and categorical groups. Appropriate preprocessing techniques were then applied, including numerical preprocessing and one-hot encoding for categorical variables.

---

**3. High-Dimensional Data After Encoding**

**Challenge:** One-hot encoding categorical variables increased the number of input features, making the dataset more complex and making direct interpretation of individual encoded features difficult.

**Technique Used:** One-hot encoding was applied to categorical variables, and feature importance values of encoded variables were subsequently consolidated back to their original business features for meaningful interpretation.

---

**4. Different Numerical Feature Scales**

**Challenge:** Numerical variables such as Age, DistanceFromHome, MonthlyIncome, and Years of Experience had different value ranges, which can affect algorithms that are sensitive to feature scale.

**Technique Used:** Numerical feature scaling was applied where required by the selected machine-learning algorithms, ensuring that variables with larger numerical ranges did not disproportionately influence the model.

---

**5. Selection of the Best Machine-Learning Model**

**Challenge:** Different classification algorithms can produce different levels of predictive performance, making it necessary to identify the most suitable model for predicting employee performance ratings.

**Technique Used:** Multiple machine-learning classification models were trained and evaluated using consistent evaluation metrics. Their performance was compared to select the most appropriate model for the project.

---

**6. Model Evaluation with Multiple Performance Metrics**

**Challenge:** Because the PerformanceRating classes were imbalanced, relying only on accuracy could provide a misleading assessment of model performance.

**Technique Used:** Multiple evaluation metrics, including Precision, Recall, F1-Score, and Confusion Matrix analysis, were used to obtain a more comprehensive evaluation of the models.

---

**7. Interpretation of Model Predictions**

**Challenge:** A predictive model alone does not explain which employee or organizational factors are most relevant to performance prediction, limiting its usefulness for business decision-making.

**Technique Used:** Feature importance analysis was performed to identify the most influential features. The encoded feature importance values were consolidated into original business features to generate actionable business insights and recommendations.
